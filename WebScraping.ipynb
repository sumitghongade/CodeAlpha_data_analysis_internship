```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from urllib.parse import urljoin
import time
```

```python
categories = {
    "Laptops": "https://webscraper.io/test-sites/e-commerce/allinone/computers/laptops",
    "Tablets": "https://webscraper.io/test-sites/e-commerce/allinone/computers/tablets",
    "Phones": "https://webscraper.io/test-sites/e-commerce/allinone/phones/touch"
}

print(categories)
```

```python
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                  "AppleWebKit/537.36 (KHTML, like Gecko) "
                  "Chrome/140.0.0.0 Safari/537.36"
}

print(headers)
```

```python
products = []

print("Product list created")

url = categories["Laptops"]

response = requests.get(url, headers=headers)

print("Status Code:", response.status_code)
```

```python
soup = BeautifulSoup(response.text, "html.parser")

print(soup.title.text)

product_list = soup.select(".thumbnail")

print("Products Found:", len(product_list))

product = product_list[0]

name = product.select_one(".title").get("title")
price = product.select_one(".price").get_text(strip=True)
description = product.select_one(".description").get_text(strip=True)

print("Product:", name)
print("Price:", price)
print("Description:", description)
```

```python
review_tag = product.select_one(".ratings .pull-right")

if review_tag:
    reviews = review_tag.get_text(strip=True)
else:
    reviews = ""

print("Reviews:", reviews)

link_tag = product.select_one(".title")

if link_tag and link_tag.get("href"):
    product_url = urljoin(url, link_tag.get("href"))
else:
    product_url = ""

print("Product URL:", product_url)
```

```python
for product in product_list:

    name = product.select_one(".title").get("title", "").strip()
    price = product.select_one(".price").get_text(strip=True)
    description = product.select_one(".description").get_text(strip=True)

    review_tag = product.select_one(".ratings .pull-right")

    if review_tag:
        reviews = review_tag.get_text(strip=True)
    else:
        reviews = ""

    link_tag = product.select_one(".title")

    if link_tag and link_tag.get("href"):
        product_url = urljoin(url, link_tag.get("href"))
    else:
        product_url = ""

    products.append({
        "Category": "Laptops",
        "Product Name": name,
        "Price": price,
        "Description": description,
        "Reviews": reviews,
        "Product URL": product_url
    })

print("Laptops collected:", len(products))
```

```python
df = pd.DataFrame(products)

print(df.head())

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print(df.columns)

df.to_csv("ecommerce_products.csv", index=False)

print("Laptops dataset saved successfully!")
```
